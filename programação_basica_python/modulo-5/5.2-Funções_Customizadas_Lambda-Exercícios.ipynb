### Q1.

- Sabendo que o código a seguir calcula o fatorial de n, escreva uma função chamada ```fatorial()``` que recebe um inteiro ```n``` como parâmetro e retorna o resultado do fatorial de ```n```. 
- No programa principal, peça ao usuário o valor de ```n```, chame a sua função e imprima o retorno.

```python
fat = 1
for i in range(1, n+1):
    fat *= i 
```

````python
## Crie aqui a função fatorial()
## lembre-se de executar a célula para que 
## a função passe a existir e possa ser chamada
def fatorial(n):
    fat = 1
    for i in range(1, n + 1):
        fat *= i
    return fat
````

````python
## Escreva aqui o programa principal e execute a célula
n_input = int(input("Digite o valor de n: "))
resultado_fatorial = fatorial(n_input)
print(f"O fatorial de {n_input} é {resultado_fatorial}")
````

### Q2.

- Escreva uma função em Python chamada ```soma_quadrados``` que recebe dois números como parâmetros e retorna a soma dos seus quadrados. 
- No programa principal solicite ao usuário que insira dois números e utilize a função para exibir a soma dos quadrados.

````python
## Crie aqui a função soma_quadrados()
def soma_quadrados(num1, num2):
    quadrado_num1 = num1 ** 2
    quadrado_num2 = num2 ** 2
    soma = quadrado_num1 + quadrado_num2
    return soma
````

````python
## Escreva aqui o programa principal e execute a célula
primeiro_numero = float(input("Digite o primeiro número: "))
segundo_numero = float(input("Digite o segundo número: "))

resultado_total = soma_quadrados(primeiro_numero, segundo_numero)
print(f"A soma dos quadrados de {primeiro_numero} e {segundo_numero} é {resultado_total}")
````

### Q3.

- Crie uma função em Python chamada ```soma_digitos``` que recebe um número inteiro como parâmetro e retorna a soma dos seus dígitos. Por exemplo, para o número 123, a função deve retornar 6, $(1 + 2 + 3)$.
    - O desafio aqui é separar os dígitos de um número inteiro usando operações aritméticas <br> <br>

- No programa principal solicite ao usuário que insira um número e utilize a função ```soma_digitos``` para calcular e exibir a soma dos seus dígitos.

````python
## Crie aqui a função soma_digitos()
def soma_digitos(numero):
    soma = 0
    while numero > 0:
        digito = numero % 10
        soma += digito
        numero //= 10
    return soma
````

````python
## Escreva aqui o programa principal e execute a célula
num_digitado = int(input("Digite um número inteiro: "))

resultado_soma = soma_digitos(num_digitado)
print(f"A soma dos dígitos de {num_digitado} é {resultado_soma}")
````

### Q4.

- Crie a função ```inverteValor()``` que recebe um inteiro de qualquer tamanho e retorna esse valor invertido usando apenas operações aritméticas
- Crie a função ```verificaInverso()``` que recebe o valor original e o valor invertido e retorna verdadeiro se ambos forem igualmente par ou igualmente ímpar. Retorne falso caso contrário.
- No programa principal, peça um valor do usuário e imprima o retorno de ambas as funções.


````python
## Crie aqui as funções inverteValor() e verificaInverso()
def inverteValor(numero_inteiro):
    valor_invertido = 0
    while numero_inteiro > 0:
        digito = numero_inteiro % 10
        valor_invertido = valor_invertido * 10 + digito
        numero_inteiro //= 10
    return valor_invertido

def verificaInverso(valor_original, valor_invertido):
    if (valor_original % 2 == 0 and valor_invertido % 2 == 0) or \
       (valor_original % 2 != 0 and valor_invertido % 2 != 0):
        return True
    else:
        return False
````

````python
## Escreva aqui o programa principal e execute a célula
num_digitado = int(input("Digite um número inteiro: "))

num_invertido = inverteValor(num_digitado)
paridade_correspondente = verificaInverso(num_digitado, num_invertido)

print(f"O valor invertido de {num_digitado} é {num_invertido}")
print(f"O número original e o invertido são igualmente pares ou ímpares? {paridade_correspondente}")
````

### Q5.

Você está desenvolvendo um programa para auxiliar em cálculos de geometria básica. Crie as seguintes funções:
- A função ```calcula_perimetro_triangulo()``` que recebe três inteiros correspondentes aos lados de um triângulo e retorna o perímetro do triângulo, ou seja, a soma dos seus lados.
- A função ```calcula_perimetro_circulo()``` que recebe um inteiro referente ao raio do círculo e retorna o perímetro do círculo, dado por $2 \pi r$. Use a constante $\pi$ da biblioteca ```math```.
- A função ```calcula_perimetro_retangulo()``` que possui um parâmetro obrigatório ```lado1``` e um opcional ```lado2```, ambos inteiros. Se o valor opcional não for fornecido, significa que se trata de um quadrado. Sua função deve calcular e retornar o perímetro do retângulo, ou seja, a soma de seus lados. 
    - Para o quadrado, é dado por $4 \times lado1$
    - Para o retângulo é dado por $2 \times lado1 + 2 \times lado2$
    
- No programa principal apresente um menu com as opções disponíveis do seu sistema e uma quarta opção ```Sair```. Solicite ao usuário a opção desejada, solicite as entradas correspondentes à opção escolhida, invoque a respective função e apresente o seu retorno. Seu programa deve retornar ao menu até que o usuário escolha a opção ```Sair```

Exemplo de interação:
```
1 - Calcular perímetro triângulo
2 - Calcular perímetro círculo
3 - Calcular perímetro retângulo
4 - Sair

Opção: 1
Digite os três lados do triângulo:
3
4
5
O perímetro é: 12

1 - Calcular perímetro triângulo
2 - Calcular perímetro círculo
3 - Calcular perímetro retângulo
4 - Sair

Opção: 3
Informe os dois lados do retângulo. Se for um quadrado, digite 0 para o segundo valor:
5
0
O perímetro é: 20

1 - Calcular perímetro triângulo
2 - Calcular perímetro círculo
3 - Calcular perímetro retângulo
4 - Sair

Opção: 4
```

````python
## Crie aqui as funções 
import math

def calcula_perimetro_triangulo(lado1, lado2, lado3):
    return lado1 + lado2 + lado3

def calcula_perimetro_circulo(raio):
    return 2 * math.pi * raio

def calcula_perimetro_retangulo(lado1, lado2=None):
    if lado2 is None:
        
        return 4 * lado1
    else:
        
        return 2 * (lado1 + lado2)

````

````python
### Escreva aqui o programa principal 
while True:
    print("\n--- Menu de Cálculos de Perímetro ---")
    print("1. Triângulo")
    print("2. Círculo")
    print("3. Retângulo / Quadrado")
    print("4. Sair")
    
    opcao = input("Escolha uma opção: ")

    if opcao == '1':
        
        l1 = int(input("Digite o comprimento do primeiro lado do triângulo: "))
        l2 = int(input("Digite o comprimento do segundo lado do triângulo: "))
        l3 = int(input("Digite o comprimento do terceiro lado do triângulo: "))
        perimetro = calcula_perimetro_triangulo(l1, l2, l3)
        print(f"O perímetro do triângulo é: {perimetro}")
        
    elif opcao == '2':
        r = int(input("Digite o raio do círculo: "))
        perimetro = calcula_perimetro_circulo(r)
        print(f"O perímetro do círculo é: {perimetro:.2f}")
        
    elif opcao == '3':
        l1 = int(input("Digite o comprimento do primeiro lado (obrigatório): "))
        entrada_lado2 = input("Digite o comprimento do segundo lado (opcional, deixe em branco para quadrado): ")
            
        if entrada_lado2 == "0":
            perimetro = calcula_perimetro_retangulo(l1)
            print(f"O perímetro do quadrado é: {perimetro}")
        else:
            l2 = int(entrada_lado2)
            perimetro = calcula_perimetro_retangulo(l1, l2)
            print(f"O perímetro do retângulo é: {perimetro}")
       
    elif opcao == '4':
        break
       
````

### Q6.

Crie uma função lambda para verificar se um número é par ou ímpar. Em seguida, solicite ao usuário um número indefinidos de valores (até que o usuário digite 0). Para cada valor de entrada, informe se é par ou ímpar.

Exemplo de interação:
```
Digite os valores que deseja verificar a paridade (digite 0 para finalizar a entrada de dados):
3
ímpar
8
par
12
par
5
ímpar
7ímpar
```

````python
verifica_paridade = lambda numero: "Par" if numero % 2 == 0 else "Ímpar"

print("Digite os valores que deseja verificar a paridade (digite 0 pra finalizar a entrada de dados): ")

while True:
    numero = int(input())

    if numero == 0:
        break
    
    resultado = verifica_paridade(numero)
    print(f"{resultado}.")
````

### Q7.

Escreva um programa que pergunte ao usuário qual operação ele deseja: maior ou menor.
Em seguida leia uma quantidade indefinida de valores do usuário, até que o usuário digite o valor zero.
Apresente ao final o maior ou menor dos valores digitados de acordo com a escolha do usuário.

**Sua solução deve incluir pelo menos uma função ```lambda```** 

Exemplo de interação:
```
Opções: (1) maior ou (2) menor?
Opção: 1

Digite os valores de entrada. Digite 0 para finalizar a entrada de valores.
4
12
15
1
0

O maior valor é: 15
```

````python
#### Escreva aqui o programa principal 
opcao_str = input("Opções: (1) maior ou (2) menor?\nOpção: ")
opcao = int(opcao_str)

comparador = None
mensagem_final_prefixo = ""

if opcao == 1:
    comparador = lambda a, b: a if a > b else b
    mensagem_final_prefixo = "O maior valor é:"
elif opcao == 2:
    comparador = lambda a, b: a if a < b else b
    mensagem_final_prefixo = "O menor valor é:"
else:
    print("Opção inválida. Escolha 1 para maior ou 2 para menor.")

primeiro_valor_recebido = False
valor_extremo = None

print("Digite os valores de entrada. Digite 0 para finalizar a entrada de valores.")

while True:
    entrada_usuario_str = input("")
    numero = int(entrada_usuario_str)

    if numero == 0:
        break
    
    if not primeiro_valor_recebido:
        valor_extremo = numero
        primeiro_valor_recebido = True
    else:
        valor_extremo = comparador(valor_extremo, numero)

if comparador is not None and primeiro_valor_recebido:
    print(f"{mensagem_final_prefixo} {valor_extremo}")
elif comparador is None:
    pass
else:
    print("Nenhum valor válido (diferente de zero) foi digitado para comparação.")
````
