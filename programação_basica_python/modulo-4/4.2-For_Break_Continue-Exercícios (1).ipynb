### Q3.

Escreva um programa que lê 10 valores inteiros positivos do usuário e ao final imprime a média dos valores digitados com duas casas decimais.

```
Digite 10 números positivos:
3
2
12
14
20
42
19
0
8
1
A média dos valores digitados é 12.10
```

````python
## Escreva e execute seu código aqui
soma = 0

for i in range(1, 11):
    valor = int(input(f"Digite o {i}º valor inteiro positivo: "))
    soma += valor
    media = soma / i

print(f"A média dos 10 valores digitados é: {media:.2f}")
````

### Q4.

Você vai criar um sistema que registra os resultados dos jogos do Atlético MG ao longo de um campeonato. Seu sistema vai receber os resultados de todos os jogos do Galo, e deve calcular a pontuação do time sabendo que vitórias valem 3 pontos, empates 1 ponto e derrotas 0 pontos.

Entrada: <br>
A primeira linha de entrada é um inteiro N com a quantidade jogos do galo. Para cada jogo você deve ler 2 inteiros, o primeiro com a quantidade de gols do galo e o segundo com a quantidade de gols do time oponente. 

Saída: <br>
Apresente a soma de vitórias, empates e derrotas do galo, junto com o cálculo da pontuação total.

Ex:
```
Entrada:
4
2
0
1
1
0
3
1
0

Saída:
Vitórias: 2
Empates: 1
Derrotas: 1
Pontuação: 7

```




````python
### Escreva e execute seu código aqui
n = int(input("Digite a quantidade de jogos: "))

soma_vitorias, soma_empates, soma_derrotas = 0, 0, 0


for jogo in range (1, n+1):
    gols_galo = int(input(f"Digite os gols do galo no jogo {jogo}: "))
    gols_op = int(input(f"Digite os gols do oponente no jogo {jogo}: "))

    if gols_galo > gols_op:
        soma_vitorias += 1
    elif gols_galo < gols_op:
        soma_derrotas += 1
    else:
        soma_empates += 1

print(f"Total de vitorias: ", soma_vitorias)
print(f"Total de empates: ", soma_empates)
print(f"Total de derrotas: ", soma_derrotas)
print(f"Pontuação: ", 3*soma_vitorias+soma_empates)
````

### Q6.

Escreva um programa que lê uma quantidade indefinida de valores e informa o maior e o menor valor digitados. A leitura é encerrada quando o usuário digitar o valor 0 (zero).

```
Entrada:
12
-1
29
11
-5
0

Saída:
Maior: 29
Menor: -5
```

````python
## Escreva e execute seu código aqui
maior = None
menor = None
primeiro_valor = True

print("Digite valores (digite 0 para encerrar):")

while True:
    valor = float(input("Digite um valor: "))
    if valor == 0:
        break
        
    if primeiro_valor:
        maior = valor
        menor = valor
        primeiro_valor = False
    else:
        if valor > maior:
            maior = valor
        
        if valor < menor:
            menor = valor
   
if primeiro_valor:
    print("Nenhum valor foi digitado.")
else:
    print(f"Maior valor: {maior:.0f}")
    print(f"Menor valor: {menor:.0f}")
````

### Q7.

Escreva um programa que lê uma quantidade indefinida de valores e informa o **produto** dos valores positivos digitados. A leitura é encerrada quando o usuário digitar o valor 0 (zero).

```
Entrada:
2
-1
1
3
-5
0

Saída:
Produto: 6
```

````python
## Escreva e execute seu código aqui

produto = 1
valores_positivos = False

print("Digite valores (digite 0 para encerrar):")

while True:
    valor = float(input("Digite um valor: "))
    if valor == 0:
        break
    
    if valor > 0:
        produto *= valor
        valores_positivos = True
                
if valores_positivos:
    print(f"O produto dos valores positivos digitados é: {produto}")
else:
    print("Nenhum valor positivo foi digitado.")
````
